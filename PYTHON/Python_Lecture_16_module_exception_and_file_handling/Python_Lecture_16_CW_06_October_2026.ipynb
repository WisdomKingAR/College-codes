```python
#exception handling
try:
    a=int(input('Enter no 1: '))
    b=int(input('Enter no 2: '))
    c=0
    c=a/b
except ZeroDivisionError:
    print('Error : Division by zero is not allowed')
except ValueError:
    print('Error : Please Enter numbers only')
else:
    print("Division Successful")
finally:
    print('End of program')
```
