```python
# accept 3 sides of triangle and tell its type
a = float(input("Enter 1st Side of the triangle : "))
b = float(input("Enter 2nd Side of the triangle : "))
c = float(input("Enter 3rd Side of the triangle : "))

if(a==b and b==c):
    print("Equilateral Triangle")
elif(a==b or b==c or a==c):
    print("Isoceles Triangle")
else:
    print("Scalene Triangle")
```

```python
#function solver

print("f(x) = ax^3 - bx^2 + cx -d            if x > k")
print("f(x) = 0                              if x = k")
print("f(x) = -ax^3 + bx^2 - cx +d           if x < k")

a=float(input("Enter value of a : "))
b=float(input("Enter value of b : "))
c=float(input("Enter value of c : "))
d=float(input("Enter value of d : "))
k=float(input("Enter value of k : "))
x=float(input("Enter value of x : "))

if(x>k):
    f = (a*(x**3))-(b*(x**2))+(c*x)-d
    print(f"Solution : {f}")
elif(x==k):
    print("Solution : 0")
else:
    f = (a*(x**3))-(b*(x**2))+(c*x)-d
    print(f"Solution : {f}")
```

```python
# BMI calculator
w = float(input("Enter weight in kg : "))
h = float(input("Enter height in m : "))

bmi = w/(h**2)

if(bmi <= 18.4):
    print(f"BMI : {bmi:.2f} Status : Underweight")
elif(bmi <= 24.9):
    print(f"BMI : {bmi:.2f} Status : Normal")
elif(bmi <= 39.9):
    print(f"BMI : {bmi:.2f} Status : Overweight")
else:
    print(f"BMI : {bmi:.2f} Status : Obese")
```

```python
#DISTANCE CONVERTER

print("Enter the value of the convereter u want to use")
print("1. Inches to Feet ")
print("2. Centimeter to Meter ")

choice = int(input("Enter the value : "))

if (choice == 1):
    inch = float(input("Enter distance in inches : "))
    feet = inch/12
    print(f"feet : {feet:.2f}")
elif(choice == 2):
    cm = float(input("Enter distance in centimeter : "))
    m = cm/100
    print(f"meter : {m:.2f}m")
else:
    print("Invalid Choice")
```
