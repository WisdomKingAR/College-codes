```python
#create datafram from the following series
s1 = pd.Series(['Python','ML','NLP','DL'])
s2 = pd.Series(['9','8','7','9'])
s3 = pd.Series(['7','8','9','9'])
df=pd.concat([s1,s2,s3],axis=1)
print(df)
```

```python
#create data frame from dictionary
data = {'Name' : ['John','Emma','Michael','Sophia'], 
        'Age' : [25,30,35,28],
        'City' : ['New York','Paris','London','Sydney']}
df = pd.DataFrame(data)
print(df)
print()
df['Scores']=pd.Series([85,90,78,92])
print(df)
print()
print(df.iloc[0])
print()
age25 = df[df["Age"]==25]
print(age25)
print()
df.drop('City',axis=1,inplace=True)
print(df)
print()
agem30 =df[df["Age"]>30]
print(agem30)
print()
df.rename(columns={'Name':'FullName'},inplace=True)
print(df)
```
