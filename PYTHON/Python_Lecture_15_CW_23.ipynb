```python
def greet():
    message = "Hello"
    print("Local",message)
greet()
message = "Namaste"
print("Global",message)
```

```python
def outer():
    message = "Local"
    def inner():
        nonlocal message
        message = "Value 1"
        print("inner :",message)
    inner()
    print("outer :",message)
outer()
```

```python
# default parameter
def greet(name,message="Hello"):
    print(message,name)
greet("Alice","おはようございます")
greet("Bob")
```

# variable length arguements
### **args(passed as tuple)** and **kwargs (passed as dictionary)**


```python
def add_all(*nums):
    return(sum(nums))
print(add_all(1,2,3,4,5,6,7,8,9))
```

```python
def student(**details):
    print(details)
student(name ="John smith" , age =20 , city ="Mumbai")
```

```python
def details(name,age=20,*skills,**info):
    print("Name :",name)
    print("Age :",age)
    print("Skills :",skills)
    print("Info :",info)
details("Alex",25,"Python","ML",city="Mumbai",year=2025)
```
