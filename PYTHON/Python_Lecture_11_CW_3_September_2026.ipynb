#Pandas

```python
import pandas as pd
```

```python
data = [10,20,30,'Jhon',50]
my_series = pd.Series(data)
print(my_series)  #default labelels 0,1,2
```

```python
a = [1,3,5]
my_series = pd.Series(a,index = ["X","Y","Z"]);
print(my_series)
```

```python
print(my_series["X"])
```

```python
grades = {"sem1":3.25,"sem2":3.28,"sem3":3.75}
my_grades = pd.Series(grades)  #We can use dictionary to give labels
print(my_grades)
```

```python
my_series = pd.Series(grades,index = ["sem1","sem2","sem4"])  #values not existing are already given NAn
print(my_series)
```

#DataFrames

```python
data = {"Name":["John","Alice","Bob"],
       "Age":[25,30,35],
       "City":["New York","London","Paris"]}
df = pd.DataFrame(data,index = [1,2,3])
print(df)
```

```python
data = [["Alice",25,"New York"],
       ["John",30,"London"],
       ["Bob",35,"Paris"]]
df = pd.DataFrame(data,columns = ["Name","Age","City"])
print(df)  #data shall be entered properly else any value is entered anywhere
```

```python
import numpy as np
data_array=  np.array([["ProductA",15.50,100],
                      ["ProductB",22.00,45],
                      ["ProductC",9.99,250]])
df = pd.DataFrame(data_array,columns = ["item","Price","stcok"])
print(df)
```

```python
data = {"Name":["John","Alice","Bob"],
       "Age":[25,30,35],
       "City":["New York","London","Paris"]}
df = pd.DataFrame(data)
df.set_index("Name",inplace = True)
print(df) #treated as index not as column
```

```python
data = {"Name":["John","Alice","Bob"],
       "Age":[25,30,35],
       "City":["New York","London","Paris"]}
df = pd.DataFrame(data,index = pd.RangeIndex(5,8,name = "index"))  # index cha name index...using range function for indices
print(df)
```

```python
print(df.iloc[2])
```

```python
data = {"Name":["John","Alice","Bob"],
       "Age":[25,30,35],
       "City":["New York","London","Paris"]}
df = pd.DataFrame(data)
print("Original df")
print(df)
#rename index
df.rename(index = {0:'A',1:'B',2:'C'},inplace = True)  #inplace = same data frame
print("Modified")
print(df)

#reset index
df.reset_index(inplace = True)
print("Reset")
print(df)
```
