```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
x_values = [0,1,2,3,4,5]
y_values = [0,1,4,9,16,25]
plt.plot(x_values,y_values)
plt.show()
```

```python
y1 = []
y2 = []
x = range(-100,100,10)
for i in x : y1.append(i**2)
for i in x : y2.append(-i**2)
plt.plot(x,y1)
plt.plot(x,y2)
plt.xlabel("x")
plt.ylabel("y")
plt.ylim(-2000,2000)
plt.axhline(0)
plt.axvline(0)
plt.title("quadratic graph")
plt.savefig("quad.png")
plt.show()
```

```python
x1=[1,2,3]
y1=[2,4,1]
plt.plot(x1,y1,label="Line 1",color='green',linestyle='dashed',linewidth=3,marker='o',markerfacecolor='blue',markersize=12)
x2=[1,2,3]
y2=[4,1,3]
plt.plot(x2,y2,label="Line 2")
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title("Two lines on same graph")
plt.legend()
plt.show()
```

```python
import seaborn as sns
```

```python
titanic = sns.load_dataset('titanic')
sns.countplot(x='pclass',data = titanic)
plt.title("Number of passengers by class")
plt.xlabel("Passenger class")
plt.ylabel("Count of passenger")
plt.show()
```

```python
ages = [2,5,70,40,30,45,50,45,43,40,444,60,7,13,57,18,90,77,32,21,20,40]
range=(0,100)
bins = 10
plt.hist(ages,bins,range,color='green',histtype='bar',rwidth=0.8)
plt.xlabel('age')
plt.ylabel('No of people')
plt.title('My histogram')
plt.show()
```

```python
ages = [2,5,70,40,30,45,50,45,43,40,444,60,7,13,57,18,90,77,32,21,20,40]
range=(0,100)
bins = 10
plt.hist(ages,bins,range,color='green',histtype='bar',rwidth=0.8,alpha=0.5)
plt.xlabel('age')
plt.ylabel('No of people')
plt.title('My histogram')
plt.show()
```

```python
x_values = [0,1,2,3,4,5]
y_values = [0,1,4,9,16,25]
plt.scatter(x_values,y_values,s=30,color="blue")
plt.show()
```

```python
x_values = [0,1,2,3,4,5]
y_values = [0,1,4,9,16,25]
plt.scatter(x_values,y_values,label="stars",marker="*",s=30,color="green")
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.title('My scatter plot')
plt.legend()
plt.show()
```

```python
activities = ['eat','sleep','work','play']
slices=[3,7,8,6]
colors =['r','y','g','b']
plt.pie(slices,labels=activities,colors=colors,startangle=90,shadow=True,explode=(0,0,0.1,0),radius=1.75,autopct="%2.2f")
plt.legend(loc = "upper right")
plt.show()
```

```python
data = pd.DataFrame({
    'Category' : ['A']*50 + ['B']*50,
    'Value' : np.random.randn(100)
})
plt.figure(figsize=(50, 30))
sns.boxplot(x='Category',y='Value',data=data)
plt.title("box plot")
plt.show()
```

```python
from sklearn.datasets import load_iris
iris = load_iris()
data = pd.DataFrame(data=iris.data, columns=iris.feature_names)
data["Species"] = iris.target
sns.pairplot(data, hue="Species")
plt.show()
```

```python
data = pd.DataFrame({
    'A':np.random.randn(100),
    'B':np.random.randn(100),
    'C':np.random.randn(100)
})
corr = data.corr()
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.title('Heatmap of correlation Matrix')
plt.show()
```
