```python
import sqlite3
conn = sqlite3.connect('college.db')
cur = conn.cursor()
cur.execute('''CREATE TABLE IF NOT EXISTS student(
roll_no INTEGER PRIMARY KEY,
name TEXT NOT NULL,
dept TEXT,
marks INTEGER)''')
print("Table Created Successfully")
```

```python
cur.execute("ALTER TABLE student ADD COLUMN email TEXT")
print("Column added successfully")
conn.commit()
conn.close()
```

```python
import sqlite3

conn = sqlite3.connect('college.db')
cur = conn.cursor()

cur.execute("INSERT INTO student (roll_no,name, dept, marks, email) VALUES (1,'abc' ,'IT', 98, 'abc@gmail.com')")
cur.execute("INSERT INTO student (roll_no,name, dept, marks, email) VALUES (2,'xyz' ,'EXTC', 88, 'xyz@gmail.com')")

conn.commit()
conn.close()
```
