```python
class Student(object):
    def __init__(self):
        self.name='Aditya'
        self.age=30
        self.marks=900
    def talk(self):
        print("Name :-",self.name)
        print("Age :-",self.age)
        print("Marks :-",self.marks)

obj=Student()
print(obj.name)
print(obj.age)
print(obj.marks)
obj.talk()
```

```python
class Student:
    def __init__(self,name,age):
        self.fullname=name
        self.age=age
    def display(self):
        print("Student FullName : %s"%self.fullname)
        print("Student Age : %d"%self.age)

s1=Student("Python",14)
s1.display()

s2=Student("Jpython",23)
s2.display()

s3=Student("Leviathan",45)
s3.display()
```

### Inheritance

```python
class Base:
    def __init__(self):
        self.name = "base class constructor"
    def display(self):
        print(self.name)
class Derived(Base):
    pass
d = Derived()
d.display()
```

### Method Overriding

```python
class Animal:
    def sound(self):
        print("Animal makes a sound")
class dog(Animal):
    def sound(self):
        print("Dog Barks")
a=Animal()
a.sound()
d=dog()
d.sound()
```

```python
class Rectangle():
    def __init__(self,length,breadth):
        self.length=length
        self.breadth=breadth
    def getArea(self):
        print(self.length*self.breadth,"is area of the rectangle")
class Square(Rectangle):
    def __init__(self,side):
        self.side=side
    def getArea(self):
        print(self.side*self.side,"is area of Square")
```

```python
s=Square(4)
r=Rectangle(4,5)
s.getArea()
r.getArea()
```

### Overloading

```python
class Person:
    def Hello(self,name=None):
        if name is not None:
            print("Hello",name)
        else:
            print("Hello")
```

```python
obj=Person() #Create instance
obj.Hello() #Call the Method
obj.Hello("Edureka")
```

```python
class Calculator:
    def multiply(self,a,b=1,c=1):
        return a*b*c
calc=Calculator()
print(calc.multiply(5))
print(calc.multiply(5,4))
print(calc.multiply(5,4,3))
```

### Operator Overloading

```python
class Test:
    def __init__(self,a):
        self.a=a
    def __add__(self, obj):
        return self.a + obj.a
t1=Test(50)
t2=Test(100)
print(t1+t2)
```

### Constructor Overloading

```python
class Base:
    def __init__(self):
        self.name = 'base class constructor'
    def display(self):
        print(self.name)
class Derived(Base):
    def __init__(self):
        pass
d = Derived()
d.display()
```

```python
class Base:
    def __init__(self):
        self.name = 'base class constructor'
    def display(self):
        print(self.name)
class Derived(Base):
    def __init__(self):
        self.name="Derived class Constructor"
d = Derived()
d.display()
```

### Super Method

```python
class Base:
    def __init__(self,nm):
        self.name = nm
    def display(self):
        print(self.name)
class Derived(Base):
    def __init__(self):
        super().__init__("Super is Accessible")
d = Derived()
d.display()
```
