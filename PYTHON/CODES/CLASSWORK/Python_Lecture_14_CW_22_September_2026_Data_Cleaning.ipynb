# Data cleaning

```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'A':[1,2,np.nan,4],
    'B':[5,np.nan,7,8],
    'C':[9,10,11,12]
})
missing_values = df.isna()
print(missing_values)
```

```python
df_filled=df.fillna(0)
print(df_filled)
```

```python
df_filled_mean = df.fillna(df.mean())
print(df_filled_mean)
```

```python
df_dropped=df.dropna()
print(df_dropped)
```

### Forward filling

```python
data = {
    'A':[1,None,None,4,None,6]
}
df = pd.DataFrame(data)
print("Before Forward Filling")
print(df)
df['A']=df['A'].fillna(method="ffill")
print("\nAfter Forward Filling")
print(df)
```

### Backward filling

```python
data = {
    'A':[1,None,None,4,None,6]
}
df = pd.DataFrame(data)
print("Before Backward Filling")
print(df)
df['A']=df['A'].fillna(method="bfill")
print("\nAfter Backward Filling")
print(df)
```

Z-score = (x-u)/d  
x = data point  
u = mean  
d = deviation  

```python
from scipy import stats
data = np.array([10,12,14,15,100,18,20])
z_scores = stats.zscore(data)
outliers = np.abs(z_scores)>2
print("Z-scores : ",z_scores)
print("Outliers : ",data[outliers])
```

```python
data = np.array([10,12,14,15,100,18,20])
Q1 = np.percentile(data,25)
Q3 = np.percentile(data,75)
IQR = Q3 - Q1
outliers = ((data < (Q1-1.5*IQR))|(data > (Q3+1.5*IQR)))
print("IQR : ",IQR)
print("Outliers : ",data[outliers])
print(Q1)
print(Q3)
```

```python
df = pd.DataFrame({
    'values' : [10,12,14,15,100,18,20]
})
Q1 = df['values'].quantile(0.25)
Q3 = df['values'].quantile(0.75)
IQR = Q3 - Q1
outliers = ((df['values'] < (Q1 - 1.5*IQR))|(df['values'] > (Q3+1.5*IQR)))
print("Outliers in DataFrame : ",df[outliers])
```
