```python
#sum of 1st n natural numbers
sum = 0
n=int(input("Enter no of elements : "))
for i in range(1,n+1):
    sum+=i
print(sum)
```

```python
#count number of even and odd numbers in the given series
p=int(input("enter p :"))
q=int(input("enter q :"))
even = 0
odd = 0
for i in range(p,q+1):
    if(i%2 == 0):
        even+=1
    else:
        odd+=1

print(f"Total Even numbers : {even}")
print(f"Total Odd numbers : {odd}")
```

```python
#prime checker
n = int(input("Enter a number : ")) 
i = 2 
count = 0 

while (i < n): 
    if (n % i == 0): 
        count += 1 
        break  
    i += 1 

if (count == 0 and n > 1):  
    print("Prime") 
else: 
    print("Not prime")
```

```python
#fibonnaci
n = int(input("Enter the last term u want in fibonacci series : "))
a=0
b=1
while(a <= n):
    print(a,end=" ")
    a,b=b,a+b
```

```python
#nested pattern loop
for i in range(1,6):
    for j in range(6-i):
        print(" ",end="")
    for k in range(2*i-1):
        print("*",end="")
    print()
```

```python
#pattern 1
for i in range(1,6):
    for j in range (6-i):
        print("*",end=" ")
    print()
```

```python
#pattern 2
for i in range(1,6):
    for j in range(1,i+1):
        print(j,end="")
    print()
```

```python
#pattern 3
n1 = 3
for i in range(n1):
    for j in range(i+1):
        print("* ",end="")
    print()

for i in range(n1):
    for j in range(i,n1-1):
        print("* ",end="")
    print()
```

```python
#fizzbuzz
for i in range(1,51):
    if(i%3==0 and i%5!=0):
        print("Fizz")
    elif(i%3!=0 and i%5==0):
        print("Buzz")
    elif(i%3==0 and i%5==0):
        print("FizzBuzz")
    else:
        print(i)
```
