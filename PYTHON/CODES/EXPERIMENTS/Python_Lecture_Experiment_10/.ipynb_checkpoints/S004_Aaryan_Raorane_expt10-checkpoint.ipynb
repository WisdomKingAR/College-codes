```python
"""
1. Implement a program to handle arithmetic exception 
"""
try:
    num1 = int(input("Enter first number: "))
    num2 = int(input("Enter second number: "))
    result = num1 / num2
    print("Result is", result)
except ZeroDivisionError:
    print("Error: You cannot divide by zero.")
```

```python
"""
2. Write a program to handle following exceptions 
a. Type Error 
b. Value Erroг 
c. Make use of else with exception and check its effect 
"""
try:
    data = input("Enter a number: ")
    number = int(data)
    result = data + number 
except ValueError:
    print("ValueError: You did not enter a valid integer.")
except TypeError:
    print("TypeError: Cannot add a string to an integer.")
else:
    print("Success: No errors occurred. Result is", result)
```

```python
"""
3. Write a program to write 1-20 numbers in a file & display same to the user.
"""
file = open("numbers.txt", "w")
for i in range(1, 21):
    file.write(str(i) + "\n")
file.close()
print("Numbers written to file. Reading now:")
file = open("numbers.txt", "r")
content = file.read()
print(content)
file.close()
```

```python
"""
4. Write a program to copy contents (your details) of one file to another file and display contents of both the files.
"""
file = open("details.txt", "w")
file.write("Name: Aarav Sharma\n")
file.write("Course: B.Tech AI & DS\n")
file.write("College: Mumbai University\n")
file.close()

source = open("details.txt", "r")
content = source.read()
source.close()

destination = open("copy_details.txt", "w")
destination.write(content)
destination.close()

print("Contents of details.txt:")
source = open("details.txt", "r")
print(source.read())
source.close()

print("Contents of copy_details.txt:")
destination = open("copy_details.txt", "r")
print(destination.read())
destination.close()
```

```python
"""
5. Write a python script to continue writing to a file until user enters choice as y and then on user input n stop writing and read the file and count words in the text file those are ending with alphabet "e".
"""
file = open("user_text.txt", "w")

choice = 'y'
while choice == 'y':
    text = input("Enter text to write: ")
    file.write(text + "\n")
    choice = input("Do you want to continue? (y/n): ")

file.close()

file = open("user_text.txt", "r")
words = file.read().split()
file.close()

count = 0
for word in words:
    if word.endswith('e'):
        count += 1

print("Number of words ending with 'e':", count)
```

```python
"""
6. Write a python script to count the number of lines from a text file which is not starting with an alphabet "T".
"""
file = open("sample.txt", "w")
file.write("This is line one\n")
file.write("Another line here\n")
file.write("Testing the script\n")
file.write("Python is fun\n")
file.close()

file = open("sample.txt", "r")
lines = file.readlines()
file.close()

count = 0
for line in lines:
    if not line.startswith('T'):
        count += 1

print("Number of lines not starting with 'T':", count)
```

```python
"""
7. Write a python script to delete the specified line giving the line number from the text file.
"""
file = open("lines.txt", "w")
file.write("Line 1\n")
file.write("Line 2\n")
file.write("Line 3\n")
file.write("Line 4\n")
file.write("Line 5\n")
file.close()

line_no = int(input("Enter line number to delete: "))

file = open("lines.txt", "r")
lines = file.readlines()
file.close()

del lines[line_no - 1]

file = open("lines.txt", "w")
file.writelines(lines)
file.close()

print("Line", line_no, "deleted successfully.")

file = open("lines.txt", "r")
print(file.read())
file.close()
```

```python
"""
8. Program to set file offsets with seek( ) and tell( ) functions in python
"""
file = open("offset_demo.txt", "w")
file.write("Hello, this is a file offset demonstration.")
file.close()

file = open("offset_demo.txt", "r")

print("Current position:", file.tell())

file.seek(7)
print("After seek(7), position:", file.tell())
print("Reading from position 7:", file.read(10))

file.seek(0)
print("After seek(0), position:", file.tell())
print("Reading from start:", file.read(5))

file.close()
```
