## Q2. Create a Vehicle class with max_speed and mileage instance attributes

```python
class Vehicle:
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed      # instance variable
        self.mileage = mileage          # instance variable

    def display(self):
        print(f"Max speed : {self.max_speed} km/h")
        print(f"Mileage   : {self.mileage} km/l")

car = Vehicle(180, 18)
bike = Vehicle(120, 45)

print("Car details")
car.display()
print()
print("Bike details")
bike.display()
```

## Q3. Create a child class Bus that will inherit all of the variables and methods of the Vehicle class

```python
class Vehicle:
    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage

class Bus(Vehicle):
    pass          # inherits everything from Vehicle

print("Class Bus created. Parent class of Bus:", Bus.__bases__[0].__name__)
```

## Q5. Create a Bus class that inherits from the Vehicle class. Give the capacity argument of Bus.seating_capacity() a default value of 50
**Expected Output:** `The seating capacity of a bus is 50 passengers`

```python
class Vehicle:
    def __init__(self, name, max_speed, mileage):
        self.name = name
        self.max_speed = max_speed
        self.mileage = mileage

    def seating_capacity(self, capacity):
        return f"The seating capacity of a {self.name} is {capacity} passengers"

class Bus(Vehicle):
    # override the parent method and give capacity a default value of 50
    def seating_capacity(self, capacity=50):
        return super().seating_capacity(capacity)

School_bus = Bus("bus", 180, 12)
print(School_bus.seating_capacity())
```

## Q6. Class Inheritance
The default fare charge of any vehicle is **seating capacity * 100**. If the Vehicle is a **Bus** instance, add an extra 10% of the total fare as a maintenance charge, so the final amount = total fare + 10% of the total fare.

**Expected Output:** `Total Bus fare is: 5500.0`

```python
class Vehicle:
    def __init__(self, name, mileage, capacity):
        self.name = name
        self.mileage = mileage
        self.capacity = capacity

    def fare(self):
        return self.capacity * 100

class Bus(Vehicle):
    def fare(self):
        amount = super().fare()
        amount += amount * 10 / 100      # extra 10% maintenance charge
        return amount

School_bus = Bus("School Volvo", 12, 50)
print("Total Bus fare is:", School_bus.fare())
```

## Q10. Create a child class for Bus class named Minibus inheriting the Bus class with a price attribute and a Print Price method (Multilevel Inheritance)

```python
class Vehicle:
    def __init__(self, name, mileage, capacity):
        self.name = name
        self.mileage = mileage
        self.capacity = capacity

    def fare(self):
        return self.capacity * 100

class Bus(Vehicle):
    pass

class Minibus(Bus):
    def __init__(self, name, mileage, capacity, price):
        super().__init__(name, mileage, capacity)   # access the parent class from inside the child
        self.price = price

    def print_price(self):
        print("Price of", self.name, "is:", self.price)

mini = Minibus("Mini Volvo", 15, 20, 2500000)
mini.print_price()
print("Total Bus fare is:", mini.fare())
print("MRO:", [c.__name__ for c in Minibus.__mro__])
```

## Q11. Create a Car class inheriting the Vehicle class, add a type attribute and print it using a display type method

```python
class Vehicle:
    def __init__(self, name, mileage, capacity):
        self.name = name
        self.mileage = mileage
        self.capacity = capacity

    def fare(self):
        return self.capacity * 100

class Car(Vehicle):
    def __init__(self, name, mileage, capacity, type):
        super().__init__(name, mileage, capacity)   # access the parent class from inside the child
        self.type = type

    def display_type(self):
        print("Type of", self.name, "is:", self.type)

c1 = Car("Honda City", 18, 5, "Sedan")
print("Type attribute:", c1.type)
c1.display_type()
print("Total fare is:", c1.fare())
```

## Q12. Method Overriding (Shape, Square and Circle)
`print(b)` calls `Shape.__str__()`, `b.fact()` calls `Shape.fact()`, `a.fact()` calls `Square.fact()` (overridden) and `b.area()` calls `Circle.area()`.

```python
class Shape:
    def __str__(self):
        return "I am a two-dimensional shape."

    def fact(self):
        return "Shapes are two-dimensional."

class Square(Shape):
    def __init__(self, length):
        self.length = length

    def fact(self):                      # overrides Shape.fact
        return "Squares have each angle equal to 90 degrees."

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

a = Square(4)
b = Circle(7)

print(b)
print(b.fact())
print(a.fact())
print(b.area())
```

## Q13. Method Overriding (Vehicle, Car and Truck)
Vehicle: speed() max speed 150, change_gear() 5 gear, show() displays Vehicle. Car overrides speed (240) and change_gear (6). Truck overrides speed (200) and change_gear (8). `truck.show()` uses the parent method.

```python
class Vehicle:
    def speed(self):
        print("Max speed 150")

    def change_gear(self):
        print("5 gear")

    def show(self):
        print("Display Vehicle")

class Car(Vehicle):
    def speed(self):
        print("Max speed 240")

    def change_gear(self):
        print("6 gear")

class Truck(Vehicle):
    def speed(self):
        print("Max speed 200")

    def change_gear(self):
        print("8 gear")

print("Truck")
truck = Truck()
truck.show()
truck.speed()
truck.change_gear()

print("\nCar")
car = Car()
car.show()
car.speed()
car.change_gear()

print("\nVehicle")
v = Vehicle()
v.show()
v.speed()
v.change_gear()
```
