# Student class with class variables and a class method

Create class with student_id and student_name as class variables

```python
class Student:
    student_id = 101                # class variable
    student_name = "Aarav Sharma"   # class variable

    @classmethod
    def display(cls):
        try:
            print(f"{'student_id':<15}{cls.student_id}")
            print(f"{'student_name':<15}{cls.student_name}")
            print(f"{'student_class':<15}{cls.student_class}")
        except AttributeError as e:
            print(f"Attribute error caught: {e}")

# student_class is not added yet, so the exception handler runs
Student.display()
```

Add a new class variable student_class and display all attributes

```python
Student.student_class = "SE A1"    # new class variable added from outside the class

Student.display()
```

# Account and Customer classes (inheritance)

```python
class Account:
    def __init__(self, number, balance):
        self.number = number
        self.balance = balance

    def deposit(self, amount):
        if amount <= 0:
            print("Deposit amount must be positive.")
        else:
            self.balance += amount
            print(f"Deposited {amount}. Balance is now {self.balance}.")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient balance, withdrawal failed.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. Balance is now {self.balance}.")


class Customer(Account):
    def __init__(self, number, balance, name, address, dob):
        super().__init__(number, balance)    # call the Account constructor
        self.name = name
        self.address = address
        self.dob = dob

    def display(self):
        print("Customer Details")
        print(f"{'Name':<17}{self.name}")
        print(f"{'Address':<17}{self.address}")
        print(f"{'DOB':<17}{self.dob}")
        print(f"{'Account Number':<17}{self.number}")
        print(f"{'Balance':<17}{self.balance}")

print("Classes Account and Customer defined successfully.")
```

Perform deposit and withdraw, then display customer details

```python
c1 = Customer(1001, 5000, "Priya Patel", "Andheri, Mumbai", "12/05/2005")

c1.deposit(2000)
c1.withdraw(1500)
print()
c1.display()
```

# Shape, Circle and Rectangle classes (method overriding)

```python
class Shape:
    def area(self, side):                 # area of a square
        return side * side


class Circle(Shape):
    def area(self, radius):               # overrides Shape.area
        return 3.14 * radius * radius


class Rectangle(Shape):
    def area(self, length, breadth):      # overrides Shape.area
        return length * breadth

print("Classes Shape, Circle and Rectangle defined successfully.")
```

Create objects and display the areas

```python
side = 4
radius = 7
length = 10
breadth = 5

s = Shape()
c = Circle()
r = Rectangle()

print(f"Area of Square    : {s.area(side):.2f}")
print(f"Area of Circle    : {c.area(radius):.2f}")
print(f"Area of Rectangle : {r.area(length, breadth):.2f}")
```

# Employee and Manager classes (inheritance)

```python
class Employee:
    def __init__(self, name, age, salary):
        self.name = name
        self.age = age
        self.salary = salary

    def display(self):
        print(f"Name   : {self.name}")
        print(f"Age    : {self.age}")
        print(f"Salary : {self.salary}")


class Manager(Employee):
    def __init__(self, name, age, salary, department):
        super().__init__(name, age, salary)   # call the Employee constructor
        self.department = department

    def displayManager(self):
        self.display()                        # reuse the parent method
        print(f"Department : {self.department}")

print("Classes Employee and Manager defined successfully.")
```

Create a Manager object and test displayManager()

```python
m1 = Manager("Rohan Mehta", 35, 85000, "Data Science")
m1.displayManager()
```
