# Create database and connection

```python
import sqlite3
from sqlite3 import Error

DB_NAME = "college.db"

try:
    conn = sqlite3.connect(DB_NAME)   # creates the file if it doesn't exist
    print(f"Connected to '{DB_NAME}' successfully.")
except Error as e:
    print(f"Connection error: {e}")
```

Create STUDENT table

```python
try:
    conn.execute("""
        CREATE TABLE IF NOT EXISTS STUDENT (
            Roll_No INTEGER PRIMARY KEY,
            Name    TEXT NOT NULL,
            Age     INTEGER CHECK (Age > 0),
            Branch  TEXT NOT NULL
        )
    """)
    conn.commit()
    print("Table STUDENT created successfully.")
except Error as e:
    print(f"Table creation error: {e}")
```

# Insert three records

```python
students = [
    (101, "Aarav Sharma", 19, "AI & DS"),
    (102, "Priya Patel",  20, "Computer"),
    (103, "Rohan Mehta",  21, "IT"),
]

try:
    conn.executemany(
        "INSERT INTO STUDENT (Roll_No, Name, Age, Branch) VALUES (?, ?, ?, ?)",
        students
    )
    conn.commit()
    print("3 records inserted successfully.")
except sqlite3.IntegrityError as e:
    print(f"Insert failed (duplicate Roll_No): {e}")
except Error as e:
    print(f"Insert error: {e}")
```

# Update age based on Roll_No

```python
roll_no = 102
new_age = 22

try:
    cur = conn.execute(
        "UPDATE STUDENT SET Age = ? WHERE Roll_No = ?",
        (new_age, roll_no)
    )
    conn.commit()
    if cur.rowcount:
        print(f"Age of Roll_No {roll_no} updated to {new_age}.")
    else:
        print(f"No student found with Roll_No {roll_no}.")
except Error as e:
    print(f"Update error: {e}")
```

# Delete a student record

```python
roll_no = 103

try:
    cur = conn.execute("DELETE FROM STUDENT WHERE Roll_No = ?", (roll_no,))
    conn.commit()
    if cur.rowcount:
        print(f"Student with Roll_No {roll_no} deleted.")
    else:
        print(f"No student found with Roll_No {roll_no}.")
except Error as e:
    print(f"Delete error: {e}")
```

# Fetch and display all records

```python
try:
    cur = conn.execute("SELECT Roll_No, Name, Age, Branch FROM STUDENT ORDER BY Roll_No")
    rows = cur.fetchall()

    print(f"{'Roll_No':<10}{'Name':<15}{'Age':<6}{'Branch'}")
    print("-" * 45)
    if not rows:
        print("(no records)")
    for r in rows:
        print(f"{r[0]:<10}{r[1]:<15}{r[2]:<6}{r[3]}")
except Error as e:
    print(f"Fetch error: {e}")
```

# Exception handling demo (duplicate Roll_No)

```python
try:
    conn.execute(
        "INSERT INTO STUDENT (Roll_No, Name, Age, Branch) VALUES (?, ?, ?, ?)",
        (101, "Duplicate Entry", 20, "IT")   
    )
    conn.commit()
except sqlite3.IntegrityError as e:
    print(f"IntegrityError caught: {e}")
except Error as e:
    print(f"Database error caught: {e}")
finally:
    conn.close()
    print("Connection closed.")
```
