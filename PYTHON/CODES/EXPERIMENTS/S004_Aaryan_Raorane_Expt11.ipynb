# Create a Vehicle class with max_speed and mileage instance attributes

Define the class with a constructor

```python
class Vehicle:
    def __init__(self, max_speed, mileage):
        self.max_speed = max_speed      # instance variable
        self.mileage = mileage          # instance variable

    def display(self):
        print(f"Max speed : {self.max_speed} km/h")
        print(f"Mileage   : {self.mileage} km/l")

print("Class Vehicle defined successfully.")
```

Create objects and display their attributes

```python
car = Vehicle(180, 18)
bike = Vehicle(120, 45)

print("Car details")
car.display()
print()
print("Bike details")
bike.display()
```
