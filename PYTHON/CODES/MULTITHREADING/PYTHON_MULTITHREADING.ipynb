# Multithreading in Python

## 1. Program using threads

```python
import threading
import time

def print_numbers():
    for i in range(1, 6):
        print("Number: " + str(i))
        time.sleep(1)

def print_letters():
    for letter in "ABCDE":
        print("Letter: " + letter)
        time.sleep(1)

start_time = time.time()

t1 = threading.Thread(target=print_numbers)
t2 = threading.Thread(target=print_letters)

t1.start()
t2.start()

t1.join()
t2.join()

end_time = time.time()
print("Total time with threads: " + str(round(end_time - start_time, 2)) + " seconds")
```

## 2. Program using sequential execution

The same two tasks are now called one after the other, with no threads.

```python
import time

def print_numbers():
    for i in range(1, 6):
        print("Number: " + str(i))
        time.sleep(1)

def print_letters():
    for letter in "ABCDE":
        print("Letter: " + letter)
        time.sleep(1)

start_time = time.time()

print_numbers()
print_letters()

end_time = time.time()
print("Total time sequentially: " + str(round(end_time - start_time, 2)) + " seconds")
```

## 3. Difference between start() and join()

```python
import threading

def show():
    print("Thread is running")

t = threading.Thread(target=show)
t.start()
t.join()
print("Main program ends after the thread")
```

## 4. Does multithreading always reduce execution time?

```python
import threading
import time

def count(n):
    total = 0
    for i in range(n):
        total += i

N = 20000000

start_time = time.time()
count(N)
count(N)
print("Sequential: " + str(round(time.time() - start_time, 2)) + " seconds")

start_time = time.time()
t1 = threading.Thread(target=count, args=(N,))
t2 = threading.Thread(target=count, args=(N,))
t1.start()
t2.start()
t1.join()
t2.join()
print("Threaded: " + str(round(time.time() - start_time, 2)) + " seconds")
```
