```python
import pandas as pd
import numpy as np
```

```python
#panda series 10 random integer find mean , access the first 5 elements through slicing
np.random.seed(0)
sr=pd.Series(np.random.randint(1,101,size=10))
print(sr)
print(np.mean(sr))
sr_slice=sr.loc[1:5]
print("\nSliced_Rows :-")
print(sr_slice)
```
