# Strip Method

```python
s="          abcabcabc      "
print(s.strip())
s="aaaaaaaabcabcabcaaaa"
print(s.strip('a'))
s="Hello world"
print(s.strip('ldoH'))
print(s.strip('ldoh'))
```

# Lstrip and Rstrip Method
### Removes only leading and trailing whitespaces by default respectively

```python
s="         Hello world      "
print(s.lstrip())
print(s.rstrip())
```

```python
"  solving any problem is an art.  ".strip('. ')
```

```python
"  solving any problem is an art.  ".strip(' ')
```

```python
str="I am Jaspreet      ."
str = str.rstrip('I am.')
str
```

```python
str="I am Jaspreet      ."
str = str.lstrip('I am.')
str
```

```python
sentence ="My name is Jaspreet  ".rstrip("my ")
sentence
```

# Split method

## syntax : string.split(separator,maxsplit)

```python
"Hell$I$am$Jaspreet".split('$',maxsplit=2)
```

```python
"Hell$I$am$Jaspreet".rsplit('$',maxsplit=2)
```

```python
"Hello world! I am tired saying this sentence again and again".split()
```

# Join method
## Syntax : separator.join(iterator)

```python
s = "Hello world! I am tired saying this sentence again and again".rsplit()
print(s)
s='    '.join(s)
print(s)
```

```python
s=s.split()
print(s)
s='_@_'.join(s)
print(s)
```

```python
s='aBc'
print(s.upper())
print(s.lower())
print(s.capitalize())
```

```python
s='Python is Fun'
print(s.replace('Py','Kr'))
print(s.count('n'))
print(len(s))
print(s.find('is'))
```
