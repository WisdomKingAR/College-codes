```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv("titanic (1).csv")
titanic.head()
```

```python
titanic.info()
```

```python
titanic['Pclass'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Passenger Class Distribution')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```

```python
titanic['Age'].plot(kind='hist', bins=20, color='skyblue', edgecolor='black')
plt.title('Age Distribution of Passengers')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
titanic['Survived'].value_counts().plot(
    kind='pie', 
    autopct='%1.1f%%', 
    colors=['salmon', 'skyblue'], 
    labels=['Not Survived', 'Survived'],
    startangle=90
)
plt.title('Proportion of Survivors and Non-Survivors')
plt.show()
```

```python
iris = pd.read_csv("Iris.csv")
iris.head()
```

```python
iris.info()
```

```python
sns.scatterplot(data=iris, x='SepalLengthCm', y='SepalWidthCm', hue='Species', palette='deep')
plt.title('Sepal Length vs. Sepal Width by Species')
plt.xlabel('Sepal Length')
plt.ylabel('Sepal Width')
plt.show()
```

```python
iris['PetalLengthCm'].plot(kind='hist', bins=15, color='skyblue', edgecolor='black')
plt.title('Petal Length Distribution')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.show()
```

```python
sns.barplot(data=iris, x='Species', y='PetalWidthCm', errorbar=None, palette='Set2')
plt.title('Average Petal Width by Species')
plt.xlabel('Species')
plt.ylabel('Average Petal Width (cm)')
plt.show()
```

```python
data_url = "http://lib.stat.cmu.edu/datasets/boston"
raw_df = pd.read_csv(data_url, sep=r"\s+", skiprows=22, header=None)
data = np.hstack([raw_df.values[::2, :], raw_df.values[1::2, :2]])
target = raw_df.values[1::2, 2]
columns = ['CRIM', 'ZN', 'INDUS', 'CHAS', 'NOX', 'RM', 'AGE', 'DIS', 'RAD', 'TAX', 'PTRATIO', 'B', 'LSTAT']
boston = pd.DataFrame(data, columns=columns)
boston['MEDV'] = target
```

```python
boston.head()
```

```python
sns.scatterplot(data=boston, x='RM', y='MEDV')
plt.show()
```

```python
sns.histplot(data=boston, x='AGE', bins=20, color='teal')
plt.title('Distribution of Units Built Before 1940')
plt.xlabel('Proportion of Owner-Occupied Units Built Prior to 1940')
plt.ylabel('Count')
plt.show()
```

```python
sns.countplot(data=boston, x='CHAS')
plt.show()
```

```python
wine = pd.read_csv("WineQT (1).csv")
wine.head()
```

```python
wine.info()
```

```python
sns.scatterplot(data=wine, x='alcohol', y='quality')
plt.title("Alcohol v/s quality")
plt.show()
```

```python
plt.hist(wine['residual sugar'])
plt.xlabel('residual sugar')
plt.ylabel('Frequency')
plt.show()
```

```python
avg_alcohol = wine.groupby('quality')['alcohol'].mean().reset_index()
sns.barplot(x='quality', y='alcohol', data=avg_alcohol, hue='quality')
plt.title('Average Alcohol Content by Wine Quality')
plt.xlabel('Wine Quality (Score)')
plt.ylabel('Average Alcohol (%)')
plt.ylim(9, 13)
plt.show()
```
