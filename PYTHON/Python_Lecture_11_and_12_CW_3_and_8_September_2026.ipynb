```python
data = {"Name":["John","Alice","Bob"],
       "Age":[25,30,35],
       "City":["New York","London","Paris"]}
df = pd.DataFrame(data)
print("Original df")
print(df)
#rename index
df.rename(index = {0:'A',1:'B',2:'C'},inplace = True)  #inplace = same data frame
print("Modified")
print(df)

#reset index
df.reset_index(inplace = True)
print("Reset")
print(df)
```

```python
data = {'Name' : ['John','Emma','Michael','Sophia','Aadi'], 
        'Age' : [25,30,35,28,69],
        'City' : ['New York','Paris','London','Sydney','Kalyan']}
df = pd.DataFrame(data)
single_row = df.loc[2] #access a single row
print("single row:")
print(single_row)
print()
row_list = df.loc[[0,3,4]] #access rows 0, 3 and 4
print("list of rows:")
print(row_list)
print()
column_list = df.loc[:,['Name', 'Age']]
print("list of cols:")
print(column_list)
print()
specific_value = df.loc[1, 'Name']
print("specific value:")
print(specific_value)
print()
```
