```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv("titanic (1).csv")
titanic.head()
```

```python
titanic.info()
```

```python
titanic['Pclass'].value_counts().plot(kind='bar', color='skyblue', edgecolor='black')
plt.title('Passenger Class Distribution')
plt.xlabel('Pclass')
plt.ylabel('Count')
plt.xticks(rotation=0)
plt.show()
```
