```python
#arithmetic operators
a=5
b=3
print(a+b)#addition
print(a-b)#substraction
print(a*b)#multiplication
print(a/b)#division
print(a%b)#modulus
print(a//b)#floor division
print(a**b)#exponent
```

```python
#conditional operators
a=5
b=3
print(a>b)#greater then
print(a<b)#less then
print(a==b)#equal to
print(a!=b)#not equal to
print(a>=b)#greater then or equal to
print(a<=b)#less then or equal to
```

```python
#bitwise operators
a=60
b=13
print(a & b)
print(a | b)
print(a ^ b)
print(~ a)
print(~ b)
print(bin(a))
print(bin(b))
```
