```python
# Q1: Leap year check
def is_leap(y):
    return (y % 4 == 0 and y % 100 != 0) or (y % 400 == 0)

for y in [2000, 1900, 2024, 2023]:
    print(y, "Leap year" if is_leap(y) else "Not a leap year")
```

```python
# Q2: Extract even numbers from a list
def ExtractEven(lst):
    return [x for x in lst if x % 2 == 0]

print(ExtractEven([1, 2, 3, 4, 5, 6]))
```

```python
# Q3: Add, sub, mul, div of two numbers
def calc(a, b):
    return a + b, a - b, a * b, (a / b if b != 0 else "Division by zero")

print(calc(20, 5))
```

```python
# Q4: Area of circle (1 argument) or triangle (2 arguments)
def area(a, b=None):
    if b is None:
        return 3.14159 * a * a
    return 0.5 * a * b

print("Circle:", area(5))
print("Triangle:", area(10, 4))
```

```python
# Q5: Largest number using variable length arguments
def largest(*nums):
    return max(nums)

print(largest(4, 9))
print(largest(4, 9, 15))
print(largest(4, 9, 15, 2, 30))
```

```python
# Q6: Demonstrate **kwargs
def show_info(**kwargs):
    for k, v in kwargs.items():
        print(k, ":", v)

show_info(name="Aaryan", branch="AI and DS", year=2)
```

```python
# Q7: Fibonacci series up to n terms using recursion
def fib(n):
    if n <= 1:
        return n
    return fib(n - 1) + fib(n - 2)

n = 10
print([fib(i) for i in range(n)])
```

```python
# Q8: Maximum value in a list using recursion
def find_max(lst):
    if len(lst) == 1:
        return lst[0]
    m = find_max(lst[1:])
    return lst[0] if lst[0] > m else m

print(find_max([3, 8, 1, 25, 7]))
```

```python
# Q9: Length of a string using recursion
def string_length(s):
    if s == "":
        return 0
    return 1 + string_length(s[1:])

print(string_length("Python"))
```

```python
# Q10: Power using recursion
def power(base, exp):
    if exp == 0:
        return 1
    return base * power(base, exp - 1)

print(power(2, 5))
```

```python
# Q11: Print star pattern in n rows using recursion
def print_pattern(n):
    if n == 0:
        return
    print_pattern(n - 1)
    print("*" * n)

print_pattern(5)
```

```python
# Q12: Factorial using recursion
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

```python
# Q13: Merge two dictionaries, overlapping keys stored as a list
def merge_dicts(d1, d2):
    result = dict(d1)
    for k, v in d2.items():
        if k in result:
            result[k] = [result[k], v]
        else:
            result[k] = v
    return result

print(merge_dicts({"a": 1, "b": 2}, {"b": 3, "c": 4}))
```

```python
# Q14: Squares of first 10 Fibonacci numbers using map
fibs = [fib(i) for i in range(10)]
print(list(map(lambda x: x * x, fibs)))
```

```python
# Q15: Lambda to add 26, and lambda to multiply x and y
add26 = lambda x: x + 26
mul = lambda x, y: x * y
print(add26(10))
print(mul(6, 7))
```

```python
# Q16: Numbers divisible by 9 using lambda
nums = [9, 12, 18, 25, 27, 40, 45, 50]
print(list(filter(lambda x: x % 9 == 0, nums)))
```

```python
# Q17 i: Filter out inactive users (no tweets)
users = [
    {"username": 'seema', "tweets": ["i love chocolate", "i am a student"]},
    {"username": 'arush', "tweets": []},
    {"username": 'kumal', "tweets": ["India", "Python"]},
    {"username": 'sam', "tweets": []},
    {"username": 'lokesh', "tweets": ["i am good"]},
]

inactive = list(filter(lambda u: not u["tweets"], users))
print(inactive)
```

```python
# Q17 ii: Inactive users, only username in uppercase
print(list(map(lambda u: u["username"].upper(), filter(lambda u: not u["tweets"], users))))
```
