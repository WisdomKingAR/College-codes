# Numpy (Numerical Python)

```python
import numpy as np
```

## Creating Numpy Array

```python
# normal array
list1=[2,4,6,8]
array = np.array(list1)
print(array)
```

```python
# array of 0 and 1
array0 = np.zeros(4)
array1 = np.ones(4)
print(array0,array1)
```

```python
# creating array with range
arr1 = np.arange(5)
arr2 = np.arange(1,6)
arr3 = np.arange(1,11,2)
print(arr1,arr2,arr3)
```

```python
# random array generator
arr = np.random.rand(5)
print(arr)
```
