# Numpy (Numerical Python)

```python
import numpy as np
```

## Creating Numpy Array and Studying its types

```python
# normal array
list1=[2,4,6,8]
array = np.array(list1)
print(array)
```

```python
# array of 0 and 1
array0 = np.zeros(4)
array1 = np.ones(4)
print(array0,array1)
```

```python
# creating array with range
arr1 = np.arange(5)
arr2 = np.arange(1,6)
arr3 = np.arange(1,11,2)
print(arr1,arr2,arr3)
```

```python
# random array generator
arr = np.random.rand(5)
print(arr)
```

```python
# random integer array generator
arr = np.random.randint(1,101,size=5)
print(arr)
```

```python
# random integer array generator with seed
np.random.seed(0)
arr = np.random.randint(1,101,size=5)
print(arr)
```

```python
# generate 5 random number uniformly distributed distributed between 0.0 and 1.0
arr = np.random.uniform(0.0,1.0,size=5)
print(arr)
```

```python
arr = np.empty(5)
print(arr)
```

```python
# 1d array
arr1d = np.zeros(5)
print(arr1d)
print(arr1d.shape)
```

```python
# 2d array
arr2d = np.zeros((2,3))
print(arr2d)
print(arr2d.shape)
```

```python
# 3d array
arr3d = np.zeros((2,3,4))
print(arr3d)
print(arr3d.shape)
```

```python
# 4d array
arr4d = np.zeros((2,3,4,5))
print(arr4d)
print(arr4d.shape)
# (batch,quantity,height,width)
```
