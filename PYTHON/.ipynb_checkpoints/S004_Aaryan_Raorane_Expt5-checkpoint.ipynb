```python
import numpy as np
```

```python
#sum of element of 3 x 3
arr3d = np.random.randint(1,11,size=(3,3))
print(arr3d)
Sum = np.sum(arr3d)
print("Sum of all Elements : ",Sum)
```

```python
#avg of element of 4 x 4
arr4d = np.random.rand(4,4)
print(arr4d)
avg = np.mean(arr4d)
print("Average of all Elements : ",avg)
```

```python
#sum of element of 2 x 3
arr2x3 = np.random.rand(2,3)
print(arr2x3)
Sum = np.sum(arr2x3[:,1])
print("Sum of all Elements of 2nd row : ",Sum)
```

```python
# find highest element in 5x5
arr5x5 = np.random.randint(1,10001,size=(5,5))
print(arr5x5)
Max = np.max(arr5x5)
print("Max among all Elements : ",Max)
```

```python
# find trace in 4x4
arr4x4 = np.random.rand(4,4)
print(arr4x4)
tr = np.trace(arr4x4)
print("sum of all diagonal elements : ",tr)
```

```python
# find highest element in 3x3
arr3x3 = np.random.rand(3,3)
print(arr3x3)
Min = np.min(arr3x3)
print("Minimum among all Elements : ",Min)
```

```python
# sum all elements > 10 in 3X3
arr3x3 = np.random.randint(1,21,size=(3,3))
print(arr3x3)
Sum = np.sum(arr3x3[arr3x3 > 10])
print("Sum of all Elements of 2nd row : ",Sum)
```

```python
# Generate a 5x5 matrix of random numbers and find the mean of the maximum values of each row.
arr5x5 = np.random.randint(1, 101, size=(5, 5))
print(arr5x5)
row_maxima = np.max(arr5x5, axis=1)
print("Row Max :", row_maxima)
mean_of_maxima = np.mean(row_maxima)
print("Mean of Max :", mean_of_maxima)
```

```python
# Create a 3x3 matrix and calculate the difference between the sum of the elements in the first row and 
# the sum of the elements in the last row.
arr3x3 = np.random.randint(1, 21, size=(3, 3))
print(arr3x3)
sum_first_row = np.sum(arr3x3[0, :])
sum_last_row = np.sum(arr3x3[-1, :])
difference = sum_first_row - sum_last_row
print("Difference between sum of first and last row : ", difference)
```

```python
# Generate a 6x6 matrix of random integers and replace all values below the median with zero.
arr6x6 = np.random.randint(1, 101, size=(6, 6))
print(arr6x6)
median_val = np.median(arr6x6)
print("Median value : ",median_val)
arr6x6[arr6x6 < median_val] = 0
print("Matrix after replacing values below median with zero:\n", arr6x6)
```

```python
# Create a 4x4 matrix and compute the sum of elements in the upper triangular part of the matrix.
arr4x4 = np.random.randint(1, 101, size=(4, 4))
print(arr4x4)
upper_tri_elements = arr4x4[np.triu_indices(4)]
sum_upper_tri = np.sum(upper_tri_elements)
print("Sum of elements in the upper triangular part : ", sum_upper_tri)
```

```python
# Generate a 3x4 matrix of random values and find the standard deviation of the elements in the third col.
arr3x4 = np.random.randint(1, 101, size=(3, 4))
print(arr3x4)
third_column_std = np.std(arr3x4[:, 2])
print("Standard deviation : ", third_column_std)
```
