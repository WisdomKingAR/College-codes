```python
a=int(input("Enter no. 1 : "))
b=int(input("Enter no. 2 : "))
c=int(input("Enter no. 3 : "))
if a>b and a>c:
    print(f"{a} is greatest")
elif a>c and b>a:
    print(f"{b} is greatest")
else:
    print(f"{c} is greatest")
```

```python
# nested if-else
a = int(input("Enter no. : "))
if a> 10:
    print("Above 10")
    if a>20:
        print("And also above 20")
    else :
        print("but not above 20")
```

# Strings and string operations

```python
a = "This is name1"
b = 'This is name2'
print(a)
print(b)
```

```python
a = "This is 'name1'"
print(a)
b = 'This is "name2"'
print(b)
```

```python
# backslash character
a = "\"hello\" This isn\'t name1"
print(a)
```

```python
# accessing string character
print(a[0])
print(a[5])
print(a[-3])
```

```python
# slicing
print(a[4:-11])
print(a[4:])
print(a[:-11])
print(a[-11:-4])
```

```python
#concatenation
x = "Helllo"
y = "World"
z = "I"
a = "am"
b = "???"
print(a + b + y + z + x)
# string repition
print(3*(a + b + y + z + x))
```

```python
# == operator
print("Hello" == "Hello")
print("Hello" == "hello")
```

```python
# != operator
print("Hello" != "Hello")
print("Hello" != "hello")
```

```python
# > operator
print("Hello" > "Hello")
print("Hello" > "hello")
```

```python
# < operator
print("Hello" < "Hello")
print("Hello" < "hello")
```

```python
s='abc'*3
print(s)
print(s[0:6:2])
print(s[0:6:2])
print(s[2:10:2])
print(s[::3])
print(s[1::3])
print(s[:8:3])

#3rd parameter as negative index(1st parameter then must be greater then 2nd parameter)

print(s[8:1:-2])
print(s[8:1:-1])
print(s[8:-1:-1])
print(s[::-1])
```
