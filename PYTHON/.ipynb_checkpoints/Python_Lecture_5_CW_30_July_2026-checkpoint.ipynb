# While Loop

```python
n = int(input("Enter n : "))
i=1
sum = 0
while i<=n:
    sum += i
    i +=1 
print(sum)
```

```python
counter = 0
while counter < 3:
    print("Inside Loop")
    counter += 1
else :
    print("Inside Else")
```

```python
n = int(input("Enter a number : ")) 
i = 2 
count = 0 

while (i < n): 
    if (n % i == 0): 
        count += 1 
        break  # Move break inside the if-condition to stop immediately if a divisor is found
    i += 1     # Increment i at the while-loop level, not inside the if-statement

if (count == 0 and n > 1):  # A prime number has 0 divisors between 2 and n-1, and must be > 1
    print("Prime") 
else: 
    print("Not prime")
```

```python
for i in range (1,51):
    if(i%2!=0):
        continue
    if(i%2==0):
        print(i,end=" ")
```

```python
c=0
while c<10:
    if c==3:
        print("using continue stateemnt : skipping iteration when count is : ",c)
        c=c+1
        continue
    if c==6:
        print("using pass statement : count is : ",c)
        c=c+1
        pass
    if c==8:
        print("using break statement : count is : ",c)
        c=c+1
        break
    print("Current count is : ",c)
    c=c+1
```

```python
# sum of digits
n = int(input("Enter n : "))
sum =0
while(n!=0) :
    sum = sum + int(n%10)
    n=int(n/10)
print(sum)
```

```python
# Reverse of a number
n = int(input("Enter n : "))
rn = 0

while n != 0:
    rn = (rn * 10) + (n % 10)
    n = n // 10  

print(rn)
```

```python
#palindrome
# Reverse of a number
n = int(input("Enter n : "))
n1 = n
rn = 0

while n != 0:
    rn = (rn * 10) + (n % 10)
    n = n // 10  # Integer division to reduce n

if (rn == n1):
    print("palindrome")
else :
    print("not palindrome")
```

```python
# fibonacci
a = 0
b = 1
n = int(input("enter no of elements in series u want : "))
for i in range (n):
    print(f"{a}",end=" ")
    a,b = b,a+b
```
