Create a list of the first 10 positive integers and print the sum of the even numbers in the list

```python
num=list(range(1,11))
```

```python
sum=0
for i in num:
    if(i%2==0):
        sum+=i
print(sum)
```

Create two dictionaries {'A': [1, 2], 'B': [3, 4]} and {'B': [5, 6], 'C': [7, 8]}, then
merge them into one dictionary and print the result.

```python
dict1 = {'A':[1,2], 'B':[3,4]}
dict2 = {'B':[5,6], 'C':[7,8]}
merged_dict = {**dict1, **dict2}
print(merged_dict)
```

For the list [3, 5, 7, 3, 5, 9, 3], create a dictionary to count the occurrences of each element and print the dictionary.

```python
elements = [3, 5, 7, 3, 5, 9, 3]
counts = {}

for num in elements:
    if num in counts:
        counts[num] += 1
    else:
        counts[num]=1
print(counts)
```
