# Input and Output in Python

```python
#by default String
num = input("Enter a number : ")
print("You entered : ", num)
print("Data type of num : ", type(num))
```

```python
#typecasting input
a=int(input("Enter first no : "))
b=int(input("Enter second no : "))
sum = a+b
print("Sum is",sum)
```

```python
# print object,seperator and end
print("GFG")
print("G","F","G",sep="#",end="@!")
print("H&M")
print("H&MMM")
```

```python
print('New Year',2026,'Never see you again',sep='.')
```

```python
print('Science '+'is '+'Elegant')
```

```python
print("Hello \t World")
print("Hello \\n World")
print(3*"Hello World")
```

```python
a,b=1,2
print(a,b)
```

```python
lst = [10,"A","hai"]
print(lst)
```

```python
d = {10:"Ram",20:"Amar"}
print(d)
```

```python
print(f"The value of a is {a} and b is {b}")
print("The value of a is {} and b is {}".format(a,b))
```
