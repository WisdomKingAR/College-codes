```python
#take a list of set return a union of all set
mylist=[{1,2,3,4},{44,58,4},{77,2,58}]
myset1=set(mylist[2]).union(set(mylist[0]).union(set(mylist[1])))
print(myset1)
```

```python
set0=set(mylist[0])
set1=set(mylist[1])
set2=set(mylist[2])
myset2=set0.union(set1.union(set2))
print(myset2)
```
