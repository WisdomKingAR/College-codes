# Tuple
### tuple is immutable

```python
tfruit=("apple","banana","chickoo","dragonfruit","grape","gauva","jackfruit","lemon")
print(tfruit)
print(type(tfruit))
```

## indexing in tuple is similar to list

```python
print(type(tfruit[0]))
print(tfruit[0])
```

```python
print(len(tfruit))
```

```python
print(tfruit[1:2])
print(tfruit[:-1])
print(tfruit[-2:])
```

<u>we can update a element in tuple by first convertin it into a list and then do changes and convert back to a tuple</u>

```python
x=tfruit
y=list(x)
y[1]="kiwi"
x=tuple(y)
print(x)
```

```python
t=('JP morgan','20','education')
print(t)
```

```python
(company,empid,profile)=t
print(company)
print(empid)
print(profile)
```
