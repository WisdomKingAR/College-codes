```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = pd.read_csv("titanic.csv")
```

```python
titanic.info()
```

```python
titanic['Age'] = titanic['Age'].fillna(titanic['Age'].median())
```

```python
titanic.info()
```

```python
from scipy import stats

titanic['z_score'] = stats.zscore(titanic['Fare'])
outliers = titanic[titanic['z_score'].abs() > 3]
outliers
```

```python
iris = pd.read_csv("Iris.csv")
```

```python
iris.info()
```

```python
q1 = iris['SepalLengthCm'].quantile(0.25)
q3 = iris['SepalLengthCm'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = iris[(iris['SepalLengthCm'] < lower_bound) | (iris['SepalLengthCm'] > upper_bound)]
print(outliers)
```

```python
iris['Z_Score'] = stats.zscore(iris['PetalWidthCm'])
outliers = iris[(iris['Z_Score'] < -3) | (iris['Z_Score'] > 3)]
print(outliers)
```

```python
from sklearn.datasets import load_boston
boston = load_boston()
df = pd.DataFrame(boston.data, columns=boston.feature_names)
df['MEDV'] = boston.target
print(df.info())
print(df.head())
```

```python
df = df.dropna(subset=['RM'])
df.info()
```

```python
q1 = df['AGE'].quantile(0.25)
q3 = df['AGE'].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr
outliers = df[(df['AGE'] < lower_bound) | (df['AGE'] > upper_bound)]
print(outliers['AGE'])
```
